##Processamento de imagem

## Módulos

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#analise de imagem
import cv2
from PIL import Image
import imutils
```

## Carregar imagens

```python
imagem = np.array(Image.open('fiap.png'))
```

## Analisar as imagens

- Ver a imagem

```python
plt.figure(figsize = (5,5)) #tamanho
plt.imshow(imagem)
plt.axis("off")
```

Rotacionar imagem

```python
angulo = 90
imagem_rotacionada = imutils.rotate(imagem, angulo)
plt.imshow(imagem_rotacionada)
plt.axis("off")
```

Mudar escala

```python
img_gc = Image.open('fiap.png').convert('LA')
```

```python
#imagem preto e branco
plt.imshow(img_gc)
#plt show()
plt.axis("off")
```

Dividir as escalas RGB

```python
# Escala R
plt.title('Escala R')
plt.ylabel('Altura {}'.format(imagem.shape[0]))
plt.xlabel('Largura {}'.format(imagem.shape[1]))
plt.imshow(imagem[ : , : , 0])
#plt.axis("off")
#plt.show()
```
