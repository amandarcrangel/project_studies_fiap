Bibliotecas

```python
#gerais
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# analises
from scipy import stats as st
import pingouin as pg
```

Dados

```python
# dados
dados_ab = pd.read_csv('exampleDataABtest.csv')
dados_ab.head()
```

```python
dados_nps = pd.read_csv('nps_example.csv', sep = ';')
dados_nps.head()
```

Teste T

comparar os grupos

```python
# estatísticas por grupo
dados_ab.groupby('group') \
        .agg(media_cliques = pd.NamedAgg('clickedTrue', 'mean'),#media (coluna, a funcao que queremos usar = mean media)
             dp_cliques = pd.NamedAgg('clickedTrue', 'std'),#desvio padrao = std
             n = pd.NamedAgg('clickedTrue', 'count')) #count
```

```python
# dividir os grupos
grA = dados_ab[dados_ab['group'] == 'A']['clickedTrue'] #separamos o grupo A e coluna clicked e gravando na váriavel grA
grB = dados_ab[dados_ab['group'] == 'B']['clickedTrue']
```

```python
# aplicar o teste
st.ttest_ind(a=grA, b=grB, alternative='two-sided') #st status tteste_ind teste t
```

```python
# se o p-value < 0,05 - então rejeito a hipótese nula
```

Usando o Pingouin

```python
# teste t USANDO PINGOUIN
pg.ttest(x=grA,
         y=grB,
         alternative='two-sided', #grupo A e diferente de B
         confidence = 0.95)
```

```python
# se o p-value < 0,05 - então rejeito a hipótese nula A foi diferente de B = B teve maior performace
```

Teste F

```python
#perfil das respostas
dados_nps.groupby('response_status') \
  .size()\
  .to_frame('n')\
  .reset_index()
```

```python
# verificar nulos
dados_nps[dados_nps['nps_score'].isnull()] #isnull existe algum null?
```

```python
# filtrar dados
dados_nps_filtrados = dados_nps[(dados_nps['response_status'] == 'Complete') & \ #apenas as pessoas que completou o questionário
  (dados_nps['nps_score'].notna())]
```

```python
# resumo dados
dados_nps_filtrados.groupby('age') \ #faixetária
                   .agg(media_nps = pd.NamedAgg('nps_score', 'mean'),
                        dp_nps = pd.NamedAgg('nps_score', 'std'),
                        n = pd.NamedAgg('nps_score', 'size')) \
                   .reset_index()
```

```python
# filtrar 75+
dados_nps_filtrados_aj = dados_nps_filtrados[dados_nps_filtrados['age'] != '75+'] #por conta de ter um número de duas pessoas apenas ter respondido não conseguimos usar então ! excluimos essas amostra
```

Usamos SCIPY

```python
# separar os grupos
dados_18_24 = dados_nps_filtrados_aj[dados_nps_filtrados_aj['age'] == '18-24']['nps_score']
dados_25_34 = dados_nps_filtrados_aj[dados_nps_filtrados_aj['age'] == '25-34']['nps_score']
dados_35_44 = dados_nps_filtrados_aj[dados_nps_filtrados_aj['age'] == '35-44']['nps_score']
dados_45_54 = dados_nps_filtrados_aj[dados_nps_filtrados_aj['age'] == '45-54']['nps_score']
dados_55_64 = dados_nps_filtrados_aj[dados_nps_filtrados_aj['age'] == '55-64']['nps_score']
dados_65_74 = dados_nps_filtrados_aj[dados_nps_filtrados_aj['age'] == '65-74']['nps_score']
```

```python
# realizar teste f
st.f_oneway(dados_18_24,
            dados_25_34,
            dados_35_44,
            dados_45_54,
            dados_55_64,
            dados_65_74)
```

```python
# se valor de p < 0,05, rejeitamos a hipótese nula
```

```python
# teste f
pg.anova(dv = 'nps_score',
         between = 'age',
         data = dados_nps_filtrados_aj,
         detailed = True)
```
